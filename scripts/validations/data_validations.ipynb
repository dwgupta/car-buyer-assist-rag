```python
# Use `%pip install` instead of `!pip install` for compatibility in notebooks
%pip install -q langchain-google-genai google-cloud-aiplatform

from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import Chroma

embeddings = GoogleGenerativeAIEmbeddings(
    model="text-embedding-005",
    vertexai=True,
    project="third-extension-474206-m2",
    location="us-central1"
)

# ChromaDB path must match the PersistentClient path (../../chroma_db)
vectorstore = Chroma(
    collection_name="car_buyer_assist",
    embedding_function=embeddings,
    persist_directory="../../chroma_db"
)
```

```python
car_buyer_assist=client.get_collection("car_buyer_assist")
car_buyer_assist.count()
```

```python
query_text="What is the fuel efficiency of Camry Hybrid?"
```

```python
query_vector = embeddings.embed_query(query_text)


```

```python
results = car_buyer_assist.query(
    query_embeddings=[query_vector],
    n_results=3,
    include=["documents", "metadatas", "distances"]
)





```

```python
# Print results in readable format
ids = results["ids"][0]
docs = results["documents"][0]
metas = results.get("metadatas", [[]])[0]
dists = results.get("distances", [[]])[0]

for i in range(len(ids)):
    print(f"{'='*60}\nResult {i+1}\n{'='*60}")
    if metas and i < len(metas) and metas[i]:
        for k, v in metas[i].items():
            print(f"  {k}: {v}")
    if dists and i < len(dists):
        print(f"  distance: {dists[i]:.4f}")
    print(f"\n{docs[i]}\n")
```
